## Advanced Recommendation System

```python
survey["PrimaryRole"] = (
    survey["DevType"]
    .str.split(";")
    .str[0]
)
```

```python
survey = survey[survey["YearsCode"] <= 40]
```

```python
survey.shape
```

```python
# Convert unknown values to NaN
survey["WorkExp"] = survey["WorkExp"].replace(
    r"(?i)^unknown$",
    np.nan,
    regex=True
)

# Convert to numeric
survey["WorkExp"] = pd.to_numeric(
    survey["WorkExp"],
    errors="coerce"
)

# Calculate median
median_workexp = survey["WorkExp"].median()

# Fill missing values with median
survey["WorkExp"] = survey["WorkExp"].fillna(median_workexp)
```

```python
survey = survey[survey["WorkExp"] <= 50]
```

```python
survey.shape
```

```python
def education_group(x):
    if pd.isna(x):
        return "Unknown"
    
    x = str(x).lower()
    
    if "doctoral" in x or "ph.d" in x:
        return "Doctorate"
    
    elif "master" in x:
        return "Master"
    
    elif "bachelor" in x:
        return "Bachelor"
    
    elif "associate" in x:
        return "Associate"
    
    elif "professional" in x:
        return "Professional"
    
    elif "secondary" in x or "high school" in x:
        return "Secondary"
    
    else:
        return "Other"
```

```python
survey["EducationGroup"] = survey["EdLevel"].apply(education_group)
```

```python
survey['OrgSize'].value_counts()
```

```python
org_mapping = {
    "Just me - I am a freelancer, sole proprietor, etc.": 0,
    "less than 20 employees": 1,
    "20 to 99 employees": 2,
    "100 to 499 employees": 3,
    "500 to 999 employees": 4,
    "1,000 to 4,999 employees": 5,
    "5,000 to 9,999 employees": 6,
    "10,000 or more employees": 7,
    "Unknown": 3,
    "i don't know": 3
}

survey["OrgSizeScore"] = survey["OrgSize"].map(org_mapping)
```

```python
survey['Employment'].value_counts()
```

```python
survey.head(1)
```

```python
survey['AISelect'].unique()
```

```python
mapping = {
    "no, and i don't plan to":0,
    "no, but i plan to soon":1,
    "yes, i use ai tools monthly or infrequently":2,
    "yes, i use ai tools weekly":3,
    "yes, i use ai tools daily":4,
    "Unknown":2
}

survey["ai_usage_score"] = survey["AISelect"].map(mapping)
```

```python
profile_features = [
    
    "PrimaryRole",
    "YearsCode",
    "WorkExp",
    "experience_group",
    "seniority",
    "EducationGroup",
    "OrgSizeScore",
    "Employment",
    "ai_usage_score"
]

developer_profile = survey[profile_features].copy()
```

```python
developer_profile.head()
```

```python
survey["LearnCode"] = survey["LearnCode"].fillna(
    survey["LearnCode"].mode()
)
```

```python
developer_profile['OrgSizeScore'] = developer_profile['OrgSizeScore'].fillna(developer_profile['OrgSizeScore'].median())
```

```python
developer_profile.info()
```

```python
developer_profile.isna().sum()
```

```python
categorical_features = [
    "PrimaryRole",
    "experience_group",
    "seniority",
    "EducationGroup",
    "Employment"
]

numerical_features = [
    "YearsCode",
    "WorkExp",
    "OrgSizeScore",
    "ai_usage_score"
]
```

```python
for col in numerical_features:
    developer_profile[col] = (
        developer_profile[col]
        .fillna(developer_profile[col].median())
    )
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)
```

```python
preprocessor
```

```python
X_profile = preprocessor.fit_transform(
    developer_profile[
        categorical_features + numerical_features
    ]
)
```

```python
profile_feature_names = (
    preprocessor
    .get_feature_names_out()
)
```

```python
print(
    "Number of profile features:",
    len(profile_feature_names)
)
```

```python
print(
    "Matrix columns:",
    X_profile.shape[1]
)
```
