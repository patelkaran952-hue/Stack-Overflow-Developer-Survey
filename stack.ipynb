## Data Cleaning

```python
# Missing values in each column
missing_count = survey.isnull().sum()
```

```python
missing_count.sort_values(ascending=False)
```

```python
missing_percent = (survey.isnull().sum() / len(survey)) * 100

missing_percent.sort_values(ascending=False)
```

```python
missing_report = pd.DataFrame({
    "Missing Count": survey.isnull().sum(),
    "Missing %": round((survey.isnull().sum() / len(survey)) * 100, 2),
    "Data Type": survey.dtypes
})

missing_report = missing_report.sort_values(
    by="Missing %",
    ascending=False
)

missing_report.head()
```

```python
empty_strings = (survey == "").sum()

empty_strings = empty_strings[empty_strings > 0]

empty_strings.sort_values(ascending=False)
```

```python
prefer_not = {}

for col in survey.columns:
    if survey[col].dtype == "object":
        count = survey[col].astype(str).str.contains(
            "Prefer",
            case=False,
            na=False
        ).sum()

        if count > 0:
            prefer_not[col] = count

pd.DataFrame.from_dict(
    prefer_not,
    orient="index",
    columns=["Count"]
).sort_values(
    by="Count",
    ascending=False
)
```

```python
missing_keywords = [
    "NA",
    "N/A",
    "None",
    "Unknown",
    "Not applicable",
    "Prefer not to say",
    "I prefer not to say"
]

for keyword in missing_keywords:
    print(f"\n{keyword}")

    for col in survey.select_dtypes(include="object"):

        count = survey[col].astype(str).str.fullmatch(
            keyword,
            case=False,
            na=False
        ).sum()

        if count > 0:
            print(f"{col}: {count}")
```

```python
survey.replace(
    [
    "NA",
    "N/A",
    "None",
    "Unknown",
    "Not applicable",
    "Prefer not to say",
    "I prefer not to say"
    ],
    np.nan,
    inplace=True
)
```

## Standardizing Data

```python
for col in categorical_cols:
    survey[col] = survey[col].str.lower()
```

```python
for col in categorical_cols:
    survey[col] = survey[col].str.strip()
```

```python
for col in categorical_cols:
    survey[col] = survey[col].str.replace(
        r"\s+",
        " ",
        regex=True
    )

# Replacing multiple spaces with a single space
```

```python
for col in categorical_cols:
    survey[col] = survey[col].str.replace(
        r"[\n\r\t]",
        "",
        regex=True
    )

# Removing invisible characters
```
