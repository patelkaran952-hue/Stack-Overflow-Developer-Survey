### Skills Features

```python
survey["num_languages"] = (
    survey["LanguageHaveWorkedWith"]
    .fillna("")
    .str.split(";")
    .apply(lambda x: len([i for i in x if i != ""]))
)

survey["num_languages"].head()
```

```python
survey["num_databases"] = (
    survey["DatabaseHaveWorkedWith"]
    .fillna("")
    .str.split(";")
    .apply(lambda x: len([i for i in x if i]))
)
```

```python
survey["num_platforms"] = (
    survey["PlatformHaveWorkedWith"]
    .fillna("")
    .str.split(";")
    .apply(lambda x: len([i for i in x if i]))
)
```

```python
survey["num_frameworks"] = (
    survey["WebframeHaveWorkedWith"]
    .fillna("")
    .str.split(";")
    .apply(lambda x: len([i for i in x if i]))
)
```

```python
survey["num_databases"] = (
    survey["DatabaseHaveWorkedWith"]
    .fillna("")
    .str.split(";")
    .apply(lambda x: len([i for i in x if i]))
)
```

### Experience Features

```python
survey["YearsCode"] = pd.to_numeric(
    survey["YearsCode"],
    errors="coerce"
)
```

```python
survey["experience_group"] = pd.cut(
    survey["YearsCode"],
    bins=[0,2,5,10,20,50],
    labels=[
        "Beginner",
        "Junior",
        "Mid",
        "Senior",
        "Expert"
    ]
)
```

```python
survey["experience_group"].value_counts()
```

### Salary Features

```python
survey["salary_bucket"] = pd.qcut(
    survey["ConvertedCompYearly"],
    q=5,
    labels=[
        "Very Low",
        "Low",
        "Medium",
        "High",
        "Very High"
    ]
)
```

```python
survey["salary_bucket"].value_counts()
```

```python
import numpy as np

survey["log_salary"] = np.log1p(
    survey["ConvertedCompYearly"]
)
```

```python
survey["has_degree"] = (
    survey["EdLevel"]
    .str.contains(
        "Bachelor|Master|Professional|Doctor",
        case=False,
        na=False
    )
    .astype(int)
)
```

### Work Style Features

```python
survey["remote"] = survey["RemoteWork"]
```

```python
survey["is_fulltime"] = (
    survey["Employment"]
    .str.contains("full-time", case=False, na=False)
    .astype(int)
)
```

### AI Features

```python
mapping = {
    "Never":0,
    "Rarely":1,
    "Sometimes":2,
    "Often":3,
    "Daily":4
}

survey["ai_usage_score"] = survey["AISelect"].map(mapping)
```

### Job Features

```python
def seniority(job):
    if pd.isna(job):
        return "Unknown"

    job = job.lower()

    if "student" in job:
        return "Student"

    elif "intern" in job:
        return "Intern"

    elif "senior" in job:
        return "Senior"

    elif "manager" in job:
        return "Manager"

    elif "lead" in job:
        return "Lead"

    else:
        return "Developer"

survey["seniority"] = survey["DevType"].apply(seniority)
```

```python
survey["is_manager"] = (
    survey["DevType"]
    .str.contains("manager", case=False, na=False)
    .astype(int)
)
```

```python
survey.to_csv("with_features.csv", index=False)
```
