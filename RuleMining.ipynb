```python
import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
pd.set_option("display.max_colwidth", 100)
```

```python
survey = pd.read_csv("with_features.csv")
```

```python
survey.head(1)
```

```python
tech = survey[['ResponseId','LanguageHaveWorkedWith']]
```

```python
tech = (
    survey["LanguageHaveWorkedWith"]
    .fillna("")
    .str.get_dummies(sep=";")
)
```

```python
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

```python
frequent_itemsets = apriori(
    tech,
    min_support=0.02,
    use_colnames=True
)

frequent_itemsets.head()
```

```python
frequent_itemsets.head()
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="lift",
    min_threshold=1
)

rules.head()
```

```python
rules.sort_values(
    "lift",
    ascending=False
).head(20)
```

```python
strong_rules = rules[
    (rules["confidence"] > 0.6) &
    (rules["lift"] > 1.5)
]

strong_rules.sort_values(
    "lift",
    ascending=False
)
```
